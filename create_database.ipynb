```python
import sqlite3
import pandas as pd
import os

# Load the cleaned dataset
csv_file = "cleaned_electricity_data.csv"

df = pd.read_csv(csv_file)

print("Dataset loaded successfully!")
print("Rows:", len(df))
print("Columns:", len(df.columns))

# Create the SQLite database
connection = sqlite3.connect("electricity.db")

# Store the dataset in a SQL table
df.to_sql(
    "electricity_data",
    connection,
    if_exists="replace",
    index=False,
    chunksize=50000
)

print("\nSQL database created successfully!")
print("Table name: electricity_data")

# Verify the table
cursor = connection.cursor()

cursor.execute("SELECT COUNT(*) FROM electricity_data")
print("Total records in SQL table:", cursor.fetchone()[0])

connection.close()
```

```python
import sqlite3
import pandas as pd

# Connect to the SQLite database
connection = sqlite3.connect("electricity.db")

# SQL query: Find the top 5 hours by average power consumption
query = """
SELECT
    Hour,
    ROUND(AVG(Global_active_power), 3) AS Avg_Power_kW
FROM electricity_data
GROUP BY Hour
ORDER BY Avg_Power_kW DESC
LIMIT 5;
"""

# Execute the query and display results
result = pd.read_sql_query(query, connection)

print("Top 5 Peak Consumption Hours:")
display(result)

connection.close()
```
